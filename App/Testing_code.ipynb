```python
import pandas as pd
data = pd.read_csv(r"D:\Lomba-Lomba\TSDN 2024\App\Data\survey lung cancer.csv")
data
```

```python
from sklearn.preprocessing import LabelEncoder

# Misalkan kolom 'gender' adalah kolom kategorikal
label_encoder = LabelEncoder()
data['GENDER'] = label_encoder.fit_transform(data['GENDER'])
data['LUNG_CANCER'] = label_encoder.fit_transform(data['LUNG_CANCER'])
```

```python
data_xxx=data[['GENDER', 'AGE', 'SMOKING', 'YELLOW_FINGERS', 'ANXIETY',
       'PEER_PRESSURE', 'CHRONIC DISEASE', 'FATIGUE ', 'ALLERGY ']]
```

```python
import os
top_100 = data_xxx.head(100)
folder_path = r"D:\Lomba-Lomba\TSDN 2024"


if not os.path.exists(folder_path):
    os.makedirs(folder_path)

# Simpan file
file_path = os.path.join(folder_path, "cancerno xxx.csv")


top_100.to_csv(file_path, index= False)

print(f"file berhasil disimpan {file_path}")
```
